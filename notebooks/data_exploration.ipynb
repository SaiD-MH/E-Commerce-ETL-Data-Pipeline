```python
import pandas as pd

df = pd.DataFrame({"invoice_date": [
        pd.Timestamp("2025-01-01 10:15:00"),
        pd.Timestamp("2025-01-02 14:30:00"),
        pd.Timestamp("2025-01-04 18:45:00"),
    ]})

df['date_key'] = df['invoice_date'].apply(lambda x: int(x.strftime("%Y%m%d")))
df['full_date'] = df['invoice_date'].apply(lambda x: x.date())
df['day_of_week'] = df['invoice_date'].apply(lambda x: x.date().weekday()+ 1)
df['day_of_week'] = df['invoice_date'].apply(lambda x: x.date().day)
df['day_name'] = df['invoice_date'].apply(lambda x: x.date().strftime("%A"))
df['week_of_year'] = df['invoice_date'].apply(lambda x: x.date().isocalendar()[1])
df['month'] = df['invoice_date'].apply(lambda x: x.date().month)
df['month_name']=df['invoice_date'].apply(lambda x: x.date().strftime("%B"))
df['quarter']=df['invoice_date'].apply(lambda x: (x.date().month - 1 )// 3 + 1)
df['year'] = df['invoice_date'].apply(lambda x: x.date().year)
df['is_weekend'] = df['invoice_date'].apply(lambda x: x.date().weekday() >= 5)
df
```
